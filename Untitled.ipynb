```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
cars_data=pd.read_csv("Cardetails.csv.xls")
cars_data
```

```python
cars_data.drop(columns=['torque'], inplace=True)
```

```python
cars_data.head()
```

```python
cars_data.isnull().sum()
```

```python
cars_data.dropna(inplace=True)
```

```python
cars_data.isnull().sum()
```

```python
cars_data.duplicated().sum()
```

```python
cars_data.drop_duplicates(inplace=True)
```

```python
cars_data.shape
```

```python
cars_data.info()
```

```python
def get_brand_name(car_name):
   car_name = car_name.split(' ')[0]
   return car_name.strip()
```

```python
def clean_data(value):
    value = value.split(' ')[0]
    value = value.strip()
    if value == '':
        value = 0
    return float(value)
```

```python
cars_data.sample(5)
```

```python
get_brand_name("Maruti Swift Dzire VDI")
```

```python
cars_data['name'] = cars_data['name'].apply(get_brand_name)
```

```python
cars_data["name"].unique()
```

```python
cars_data["mileage"]=cars_data['mileage'].apply(clean_data)
```

```python
cars_data["max_power"]=cars_data["max_power"].apply(clean_data)
```

```python
cars_data["engine"]=cars_data["engine"].apply(clean_data)
```

```python
cars_data.info()
```

```python
cars_data.head()
```

```python
from sklearn.preprocessing import LabelEncoder

# Create LabelEncoder object
le = LabelEncoder()

# List of categorical columns
cat_cols = ['name', 'fuel', 'seller_type', 'transmission', 'owner']

# Apply Label Encoding on each column
for col in cat_cols:
    cars_data[col] = le.fit_transform(cars_data[col])

# Display first 5 rows
cars_data.head()
```

```python
cars_data.info()
```

```python
input_data=cars_data.drop(columns=["selling_price"])
output_data=cars_data["selling_price"]
```

```python
X_train,X_test,y_train,y_test=train_test_split(input_data,output_data,test_size=0.2)
```

```python
model=LinearRegression()
```

```python
model.fit(X_train,y_train)
```

```python
predict=model.predict(X_test)
```

```python
print(predict)
```
